# Redox FHIR Pipeline - Configuration

This notebook sets up the configuration for the Redox FHIR ingestion pipeline.

## Prerequisites
- Databricks workspace with Unity Catalog enabled
- A catalog and schema created for FHIR data
- Redox FHIR bundles available in a mounted location

## Configuration Parameters
| Parameter | Description |
|-----------|-------------|
| `catalog_use` | Unity Catalog name for FHIR data |
| `schema_use` | Schema name within the catalog |
| `mount_path` | Path to mounted Redox FHIR files |

_Note: Attach this notebook to a Serverless SQL Warehouse for execution._

```sql
-- ============================================================================
-- CONFIGURATION: Update these values for your environment
-- ============================================================================

-- Catalog and Schema
DECLARE OR REPLACE VARIABLE catalog_use STRING DEFAULT 'redox_fhir';
DECLARE OR REPLACE VARIABLE schema_use STRING DEFAULT 'bronze';

-- Mount path where Redox FHIR JSON files are stored
-- Example: /mnt/redox/fhir/ or /mnt/landing/redox/
DECLARE OR REPLACE VARIABLE mount_path STRING DEFAULT '/mnt/redox/fhir/';
```

```sql
-- Override with job parameters if provided
SET VARIABLE catalog_use = COALESCE(:catalog_use, catalog_use);
SET VARIABLE schema_use = COALESCE(:schema_use, schema_use);
SET VARIABLE mount_path = COALESCE(:mount_path, mount_path);
```

```sql
-- Set the active catalog and schema
USE IDENTIFIER(catalog_use || '.' || schema_use);
```

```sql
-- Verify configuration
SELECT 
  current_catalog() AS catalog,
  current_schema() AS schema,
  mount_path AS fhir_source_path;
```

```sql
-- List files in the mount to verify access
DECLARE OR REPLACE VARIABLE list_stmt STRING;
SET VARIABLE list_stmt = "LIST '" || mount_path || "' LIMIT 10;";
EXECUTE IMMEDIATE list_stmt;
```
